### About the Outputs

If a Colab Notebook is saved after a Code Cell executes, the output will also be saved with the Notebook.  **This means that the presence of output in a code cell does not necessarily indicate that the code was run during the current session (since the Notebook was last loaded).**  While the output is saved, the value of any variables used in the code are not.  All variables will be reset to undefined when a Colab Notebook loads.

If cells further down in the Notebook depend on variables set in an earlier code cell, those later cells will not run correctly until the previous cell is run.  **Always look for the checkmark**, and do not rely on just seeing the code output.

**Scroll past the code and notice that output for the cell is already there, even though you didn't run the code.**  As far what is actually being graphed here, this is a Sigmoid function which is an example of an activation function in a neural networks.  While the exact form of the Sigmoid isn't that important for our class, the fact that it is not a linear function is actually critical for our neural networks.  We'll talk more about this next week.



````python
import numpy as np
import matplotlib.pyplot as plt

def plot_sigmoid():
    """
    Creates a visualization of the sigmoid activation function.
    The sigmoid function is defined as f(x) = 1 / (1 + e^(-x))
    """

    # Generate x values from -10 to 10
    x = np.linspace(-10, 10, 200)

    # Calculate sigmoid values
    sigmoid = 1 / (1 + np.exp(-x))

    # Create the plot
    plt.figure(figsize=(10, 6))
    plt.plot(x, sigmoid, 'b-', linewidth=2, label='Sigmoid')

    # Add a horizontal line at y=0.5
    plt.axhline(y=0.5, color='r', linestyle='--', alpha=0.3)

    # Add a vertical line at x=0
    plt.axvline(x=0, color='r', linestyle='--', alpha=0.3)

    # Customize the plot
    plt.grid(True, alpha=0.3)
    plt.title('Sigmoid Activation Function', pad=15)
    plt.xlabel('x')
    plt.ylabel('σ(x)')

    # Add legend
    plt.legend()

    # Set axis limits for better visualization
    plt.xlim(-10, 10)
    plt.ylim(-0.1, 1.1)

    # Add text annotation explaining the function
    plt.text(3, 0.2, r'$\sigma(x) = \frac{1}{1 + e^{-x}}$', fontsize=12, bbox=dict(facecolor='white', alpha=0.7))

    plt.tight_layout()
    plt.show()

# Execute the function
plot_sigmoid()
````

### Saving Ouputs

When you run code in a Notebook and attempt to close it, Google will prompt you to save your work. You can choose to save the Notebook to a folder in your own Google Drive.

The main advantage of saving is that your saved Notebook will include any outputs from the Code Cells you have already run. This can be useful if you want to revisit or review your results later without rerunning the code.

If you choose not to save the Notebook, keep in mind that when you restart from the original Homework copy, all outputs will be cleared, and you will need to rerun the Code Cells to regenerate them.

### Restarting Troublesome Notebooks

If a Notebook does not appear to be running correctly, go to the "Runtime" menu and select "Restart Session".  This will completely clear the state of all variables and give you a fresh start.  Go back through the Code Cells in the Notebook starting at the top.  Execute each Code Cell one at a time.  As we'll see in the next Notebook, some of the Code Cells will create user interface elements such as Sliders, Text Fields, or Buttons which you will need to interact with before moving to the next Code Cell.  Once you're sure a Code Cell has run to completion and you have carried out whatever interaction is needed with any User Interface elements created by that Code Cell, move on to the next and execute.  This systematic approach ensures that all dependencies are properly initialized and the Notebook functions as expected.

### Timeouts

If you leave a Colab notebook idle for too long, Google will disconnect you from the processor running your code.  While you can continue to read the text and study any outputs you generated before getting disconnected, you’ll need to restart the notebook if you want to run code again.

Typically, you cannot start by running a code cell in the middle of the notebook, as many cells depend on results generated by earlier cells.

When restarting, remember that code cells must be run in order, starting at the top. In some cases you can use the *Runtime* menu on the menu bar to access options like *Run all*, which executes the entire notebook, or *Run before*, which runs all cells above the currently selected cell.  Do keep in mind that **for some of our pre-configured Notebooks, you will need to interact with user interface elements generated by earlier Code Cells before later Code Cells will run correctly, so in some cases you will need to manually execute each cell in order**.
